## Рішення
Маємо функцію попиту:


$$
Q(x) = 2000 - 0.8x
$$



Собівартість одних навушників: 800 грн.  
Функція прибутку:


$$
P(x) = (x - 800)(2000 - 0.8x)
$$



### Крок 1. Розкриємо дужки


$$
P(x) = -0.8x^2 + 2640x - 1{,}600{,}000
$$



### Крок 2. Знайдемо похідну


$$
P'(x) = -1.6x + 2640
$$



### Крок 3. Критична точка


$$
P'(x) = 0 \quad \Rightarrow \quad x^* = \frac{2640}{1.6} = 1650
$$



### Крок 4. Друга похідна


$$
P''(x) = -1.6 < 0 \quad \Rightarrow \text{точка максимуму.}
$$



### Крок 5. Максимальний прибуток


$$
P(1650) = (1650-800)(2000-0.8\cdot1650) = 850 \cdot 680 = 578{,}000
$$



**Висновок:**  
Оптимальна ціна = **1650 грн**  
Максимальний прибуток = **578 000 грн**


```python
import numpy as np
from scipy.optimize import minimize_scalar

# Функція прибутку
def P(x):
    return (x-800)*(2000-0.8*x)

# Аналітичне рішення
x_star = 2640/1.6
profit_star = P(x_star)
print("Аналітичне рішення:")
print("Оптимальна ціна:", x_star)
print("Максимальний прибуток:", profit_star)

# Чисельне рішення
res = minimize_scalar(lambda x: -P(x), bounds=(800,2500), method='bounded')
print("\nЧисельне рішення:")
print("Оптимальна ціна:", res.x)
print("Максимальний прибуток:", P(res.x))
```

$$

2x+y=6, \quad x+2y=9 \quad \Rightarrow \quad (x^*,y^*)=(3,3)

# Задача 3. Оптимальний план виробництва





## Умови

Меблева фабрика виготовляє два види продукції: стільці та столи.  
Ресурси:

| Ресурс        | Стілець | Стіл | Запас |
|---------------|---------|------|-------|
| Деревина (м²) | 2       | 4    | 120   |
| Час (год)     | 3       | 2    | 90    |

Прибуток:  
- Стілець = 500 грн  
- Стіл = 800 грн  

### Формулювання задачі
Позначимо:
- \(x\) — кількість стільців  
- \(y\) — кількість столів  

### Рішення

Цільова функція:


$$
P(x,y) = 500x + 800y
$$



Обмеження:


$$
2x + 4y \leq 120
$$




$$
3x + 2y \leq 90
$$




$$
x, y \geq 0
$$



### Аналітичний розв'язок
Перевіримо вузлові точки системи обмежень:

1. Якщо \(x=0\):  


$$
4y \leq 120 \Rightarrow y \leq 30
$$




$$
2y \leq 90 \Rightarrow y \leq 45
$$


→ максимум при \(y=30\), прибуток = \(800 \cdot 30 = 24{,}000\).

2. Якщо \(y=0\):  


$$
2x \leq 120 \Rightarrow x \leq 60
$$




$$
3x \leq 90 \Rightarrow x \leq 30
$$


→ максимум при $$(x=30),$$ прибуток = $$(500 \cdot 30 = 15{,}000).$$

3. Точка перетину:


$$
2x+4y=120, \quad 3x+2y=90
$$


Розв'язок: $$(x=10, y=25).$$  
Прибуток:


$$
P(10,25) = 500\cdot10 + 800\cdot25 = 30{,}000
$$



### Використання ресурсів
- Деревина: $$(2\cdot10 + 4\cdot25 = 120)$$ повне використання.  
- Час: $$(3\cdot10 + 2\cdot25 = 80)$$ залишок 10 годин.

**Висновок:**  
Оптимальний план: \(x=10\) стільців, \(y=25\) столів.  
Максимальний прибуток = **30 000 грн**.


```python
from scipy.optimize import linprog

c = [-500,-800]
A = [[2,4],[3,2]]
b = [120,90]

res = linprog(c, A_ub=A, b_ub=b, bounds=(0,None))
print("Оптимальний план:", res.x)
print("Максимальний прибуток:", -res.fun)
```

# Задача 4. Комплексний аналіз виробництва

## Умови
Це фіналізуюча задача дисципліни, яка поєднує інструменти з різних тем: диференціювання, інтегрування, системи рівнянь та оптимізацію.

Завод виробляє продукцію протягом 10-годинної зміни.

Продуктивність (одиниць на годину) змінюється протягом дня за формулою:

P(t)=100+40t−4t2

де t — час у годинах від початку зміни 0≤t≤10.

Вартість виробництва однієї одиниці залежить від двох параметрів: витрат на сировину x та витрат на енергію y (грн):

C(x,y)=x
2
 +y
2
 −10x−8y+50
Початкові значення параметрів визначаються з бюджетних обмежень:

$$
\begin{cases}
2x + y = 20 \\
x + 3y = 25
\end{cases}
$$

## Рішення

### Частина перша. Продуктивність

Функція продуктивності:
$$
P(t) = 100 + 40t - 4t^2, \quad 0 \leq t \leq 10
$$
1. Похідна:
$$
P'(t) = 40 - 8t
$$
- при \(t=2\): \(P'(2)=24 > 0\) → продуктивність зростає  
- при \(t=5\): \(P'(5)=0\) → точка максимуму  
- при \(t=8\): \(P'(8)=-24 < 0\) → продуктивність спадає  

2. Момент пікової продуктивності:
$$
t^* = 5, \quad P(5) = 200
$$
3. Загальний обсяг виробництва:
$$
\int_0^{10} P(t)\,dt = \int_0^{10} (100+40t-4t^2)\,dt
$$

$$
= [100t + 20t^2 - \tfrac{4}{3}t^3]_0^{10} = 1000 + 2000 - 1333.3 \approx 1666.7
$$

### Частина друга. Оптимізація витрат
Функція витрат:
$$
C(x,y) = x^2 + y^2 - 10x - 8y + 50
$$
Система бюджетних обмежень:
$$
\begin{cases}
2x + y = 20 \\
x + 3y = 25
\end{cases}
$$
Розв'язок системи:
$$
(x_0,y_0) = (5,10)
$$
Мінімізація функції витрат → оптимальні параметри:
$$
(x^*,y^*) \approx (5,4), \quad C(x^*,y^*) = 9
$$
---
Загальна вартість:
$$
1666.7 \cdot 9 \approx 15{,}000
$$



**Висновок:**  
- Пік продуктивності: \(t=5\), \(P(5)=200\)  
- Загальний обсяг виробництва: ≈1667 одиниць  
- Мінімальна вартість одиниці: 9 грн  
- Загальна вартість виробництва: ≈35 000 грн



```python
import numpy as np
from scipy.optimize import minimize, minimize_scalar
from scipy.integrate import quad

# Функція продуктивності
def P(t):
    return 100 + 40*t - 4*t**2

# Перевірка похідної у точках
print("P'(2) =", 40-8*2)
print("P'(5) =", 40-8*5)
print("P'(8) =", 40-8*8)

# Пошук піку продуктивності
res_peak = minimize_scalar(lambda t: -P(t), bounds=(0,10), method='bounded')
print("Пік продуктивності: t* =", res_peak.x, ", P(t*) =", P(res_peak.x))

# Загальний обсяг виробництва
total_prod, _ = quad(P, 0, 10)
print("Загальний обсяг виробництва:", total_prod)

# Функція витрат
def C(v):
    x,y=v
    return x**2 + y**2 - 10*x - 8*y + 50

# Початкові параметри з системи рівнянь
# 2x+y=20, x+3y=25
# Розв'язок: x=5, y=10
x0, y0 = 5, 10

# Мінімізація витрат
res_cost = minimize(C, [x0,y0], method='BFGS')
print("Оптимальні параметри:", res_cost.x)
print("Мінімальна вартість одиниці:", res_cost.fun)

# Загальна вартість
print("Загальна вартість виробництва:", total_prod*res_cost.fun)

```
