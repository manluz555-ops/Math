# Аналітичне рішення

Маємо функцію:


\[
u(x,y) = x^2 + y^2, \quad F(u) = u^3 \cos u, \quad f(x,y) = F(u(x,y)).
\]



---

## Крок 1. Обчислимо похідні


\[
f(x,y) = (x^2+y^2)^3 \cos(x^2+y^2).
\]



Позначимо \(u = x^2+y^2\). Тоді:


\[
\frac{\partial f}{\partial x} = \frac{dF}{du} \cdot \frac{\partial u}{\partial x}, \quad
\frac{\partial f}{\partial y} = \frac{dF}{du} \cdot \frac{\partial u}{\partial y}.
\]



---

## Крок 2. Знайдемо похідну \(F(u)\)


\[
F(u) = u^3 \cos u,
\]




\[
F'(u) = 3u^2 \cos u - u^3 \sin u.
\]



---

## Крок 3. Підставимо похідні \(u\)


\[
\frac{\partial u}{\partial x} = 2x, \quad \frac{\partial u}{\partial y} = 2y.
\]



---

## Крок 4. Остаточний вираз для градієнта


\[
\nabla f(x,y) = \Big( (3u^2 \cos u - u^3 \sin u)\cdot 2x,\; (3u^2 \cos u - u^3 \sin u)\cdot 2y \Big),
\]


де \(u = x^2+y^2\).

---

## Крок 5. Обчислимо у точці (1,1)


\[
u = 1^2 + 1^2 = 2,
\]




\[
\nabla f(1,1) = \big( (3\cdot 2^2 \cos 2 - 2^3 \sin 2)\cdot 2,\; (3\cdot 2^2 \cos 2 - 2^3 \sin 2)\cdot 2 \big).
\]



Спрощуємо:


\[
\nabla f(1,1) = (24 \cos 2 - 16 \sin 2,\; 24 \cos 2 - 16 \sin 2).
\]



---

## Крок 6. Числове значення


\[
\nabla f(1,1) \approx (-2.61,\; -2.61).
\]




```python
# --- Крок 1. Імпорт бібліотек ---
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# --- Крок 2. Оголошення змінних ---
x, y = sp.symbols('x y')

# --- Крок 3. Визначення функції ---
u = x**2 + y**2
F = u**3 * sp.cos(u)
f = F

# --- Крок 4. Обчислення градієнта ---
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)
gradient = (df_dx, df_dy)

print("∇f(x,y) =", gradient)

# --- Крок 5. Обчислення у точці (1,1) ---
grad_at_point = (df_dx.subs({x:1, y:1}), df_dy.subs({x:1, y:1}))
print("∇f(1,1) =", grad_at_point)

# --- Крок 6. Побудова графіка ---
# Перетворимо функцію у числову
f_num = sp.lambdify((x,y), f, 'numpy')

X = np.linspace(-2,2,100)
Y = np.linspace(-2,2,100)
X, Y = np.meshgrid(X,Y)
Z = f_num(X,Y)

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)

# Відобразимо точку (1,1) та вектор градієнта
ax.quiver(1, 1, f_num(1,1),
          float(grad_at_point[0]), float(grad_at_point[1]), 0,
          color='red', length=0.5, normalize=True)

ax.set_title("Поверхня f(x,y) та градієнт у точці (1,1)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x,y)")
plt.show()
```

```python
# --- Крок 1. Імпорт бібліотек ---
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# --- Крок 2. Оголошення змінних ---
x, y = sp.symbols('x y')

# --- Крок 3. Визначення функції ---
u = x**2 + y**2
f = u**3 * sp.cos(u)

# --- Крок 4. Обчислення градієнта ---
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

# --- Крок 5. Числове значення у точці (1,1) ---
grad_at_point = (df_dx.subs({x:1, y:1}), df_dy.subs({x:1, y:1}))
grad_numeric = (float(grad_at_point[0]), float(grad_at_point[1]))

print("Градієнт у точці (1,1):", grad_numeric)

# --- Крок 6. Побудова контурного графіка ---
f_num = sp.lambdify((x,y), f, 'numpy')

X = np.linspace(-2,2,200)
Y = np.linspace(-2,2,200)
X, Y = np.meshgrid(X,Y)
Z = f_num(X,Y)

plt.figure(figsize=(8,6))
contours = plt.contour(X, Y, Z, 20, cmap='viridis')
plt.clabel(contours, inline=True, fontsize=8)

# Відобразимо точку (1,1) та вектор градієнта
plt.quiver(1, 1, grad_numeric[0], grad_numeric[1],
           color='red', angles='xy', scale_units='xy', scale=1)

plt.title("Контурний графік f(x,y) та градієнт у точці (1,1)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
