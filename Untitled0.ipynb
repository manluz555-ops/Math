# Задача 1. Аналіз рейтингів мобільного додатку



**Дані:**
- Всього відгуків: 2000
- Розподіл:
  - 1⭐ → 120
  - 2⭐ → 180
  - 3⭐ → 320
  - 4⭐ → 580
  - 5⭐ → 800

**PMF:**

P(1) = 120/2000 = 0.06  
P(2) = 180/2000 = 0.09  
P(3) = 320/2000 = 0.16  
P(4) = 580/2000 = 0.29  
P(5) = 800/2000 = 0.40  

**CDF:**

F(1) = 0.06  
F(2) = 0.15  
F(3) = 0.31  
F(4) = 0.60  
F(5) = 1.00  

**Математичне сподівання:**
E[X] = 3.88  

**Дисперсія та стандартне відхилення:**
Var(X) = 1.45  
σ = 1.20  

**Ймовірності:**
- Негативна оцінка (1 або 2): P(X ≤ 2) = 0.15  
- Висока оцінка (4 або 5): P(X ≥ 4) = 0.69  

**Медіана:**
Медіана = 4⭐



```python
import numpy as np
import pandas as pd

# Дані
ratings = np.array([1, 2, 3, 4, 5])
counts = np.array([120, 180, 320, 580, 800])
total = counts.sum()

# PMF
pmf = counts / total

# CDF
cdf = np.cumsum(pmf)

# Таблиця
table = pd.DataFrame({
    "Оцінка": ratings,
    "PMF": pmf,
    "CDF": cdf
})
print(table)

# Математичне сподівання
E = np.sum(ratings * pmf)

# Дисперсія та стандартне відхилення
E2 = np.sum((ratings**2) * pmf)
Var = E2 - E**2
sigma = np.sqrt(Var)

print("E[X] =", E)
print("Var(X) =", Var)
print("σ =", sigma)

# Ймовірності
p_negative = cdf[1]   # P(X ≤ 2)
p_high = pmf[3] + pmf[4]  # P(X ≥ 4)

print("P(негативна оцінка) =", p_negative)
print("P(висока оцінка) =", p_high)

# Медіана
median = ratings[np.where(cdf >= 0.5)[0][0]]
print("Медіана =", median)

```

# Задача 2. Аналіз доставки замовлень



Ви аналізуєте роботу служби доставки інтернет-магазину. За статистикою, кожне замовлення доставляється вчасно з ймовірністю 0.88. Сьогодні відправлено 50 замовлень.

Кількість вчасно доставлених замовлень X описується біноміальним розподілом
X∼Binomial(n=50,p=0.88).

**Завдання**

Обчисліть математичне сподівання E[X] та стандартне відхилення
σ кількості вчасно доставлених замовлень.

Яка ймовірність, що всі 50 замовлень доставлять вчасно?

Яка ймовірність, що рівно 45 замовлень доставлять вчасно?

Яка ймовірність, що вчасно доставлять від 42 до 46 замовлень?

Яка ймовірність, що більше 5 замовлень запізняться?

Використайте scipy.stats.binom для обчислень.

**Рішення**

Маємо: X ~ Binomial(n=50, p=0.88)

1. Математичне сподівання
E[X] = n·p = 50·0.88 = 44

2. Стандартне відхилення
σ = √(n·p·(1-p)) = √(50·0.88·0.12) = √(5.28) ≈ 2.30

3. Ймовірність, що всі 50 замовлень вчасно
P(X=50) = (0.88)^50 ≈ 0.003

4. Ймовірність, що рівно 45 замовлень вчасно
P(X=45) = C(50,45)·(0.88)^45·(0.12)^5 ≈ 0.17

5. Ймовірність, що 42 ≤ X ≤ 46
P(42 ≤ X ≤ 46) = F(46) - F(41) ≈ 0.63

6. Ймовірність, що більше 5 замовлень запізняться
Запізнення = 50 - X
Більше 5 запізнень → X < 45
P(X<45) = F(44) ≈ 0.55




```python
from scipy.stats import binom
import numpy as np

n, p = 50, 0.88

E = n*p
sigma = np.sqrt(n*p*(1-p))

print("E[X] =", E)
print("σ =", sigma)

print("P(X=50) =", binom.pmf(50, n, p))
print("P(X=45) =", binom.pmf(45, n, p))
print("P(42≤X≤46) =", binom.cdf(46, n, p) - binom.cdf(41, n, p))
print("P(X<45) =", binom.cdf(44, n, p))
```

# Задача 3. Час між замовленнями в інтернет-магазині

**Умова**

Ви аналізуєте потік замовлень в інтернет-магазині. В середньому надходить 6 замовлень на годину, тобто одне замовлення кожні 10 хвилин.  

Час між замовленнями  T описується показниковим розподілом з параметром

λ=6 замовлень на годину або

λ=0.1 замовлень на хвилину.

Пам'ятаємо, що для показникового розподілу в scipy.stats.expon параметр scale — це середній час між подіями:
*scale*=1/λ=10 хвилин.



**Завдання**

Обчисліть математичне сподівання E[T] та стандартне відхилення
σ часу між замовленнями.

Яка ймовірність, що наступне замовлення надійде менше ніж за 5 хвилин?

Яка ймовірність, що доведеться чекати більше 15 хвилин?

Яка ймовірність, що замовлення надійде в проміжку від 5 до 15 хвилин?

Знайдіть медіанний час очікування — час, за який з ймовірністю 50% надійде наступне замовлення.

Використайте scipy.stats.expon для обчислень.

**Рішення**

λ = 0.1 замовлень/хв, scale = 10 хв

1. Математичне сподівання

E[T] = 1/λ = 10 хв

2. Стандартне відхилення
σ = 1/λ = 10 хв

3. Ймовірність, що T < 5
F(5) = 1 - e^(-5/10) = 1 - e^(-0.5) ≈ 0.393

4. Ймовірність, що T > 15
P(T>15) = e^(-15/10) = e^(-1.5) ≈ 0.223

5. Ймовірність, що 5 < T < 15
F(15) - F(5) = (1 - e^(-1.5)) - (1 - e^(-0.5)) = e^(-0.5) - e^(-1.5) ≈ 0.170

6. Медіана
F(t)=0.5 → 1 - e^(-t/10) = 0.5 → e^(-t/10)=0.5 → t=10·ln2 ≈ 6.93 хв


```python
from scipy.stats import expon

scale = 10

E = scale
sigma = scale

print("E[T] =", E)
print("σ =", sigma)

print("P(T<5) =", expon.cdf(5, scale=scale))
print("P(T>15) =", 1 - expon.cdf(15, scale=scale))
print("P(5<T<15) =", expon.cdf(15, scale=scale) - expon.cdf(5, scale=scale))
print("Медіана =", expon.ppf(0.5, scale=scale))
```

# Задача 4. Перевірка середнього балу студентів

**Умова**

Минулого року середній бал студентів на іспиті з математики становив 72 бали. Цього року викладач змінив методику підготовки і хоче перевірити, чи вплинуло це на результати.



Зібрано результати 35 випадкових студентів:

Вибіркове середнє:

xˉ=76.8 балів

Вибіркове стандартне відхилення:

s=14.2 балів

Розмір вибірки:

n=35


**Завдання**

Сформулюйте нульову гіпотезу H0 та альтернативну гіпотезу H1.

Обчисліть стандартну похибку середнього.

Обчисліть t-статистику

Знайдіть p-value для двостороннього тесту, використовуючи scipy.stats.t.cdf.

Побудуйте 95% довірчий інтервал для середнього балу. Використовуй те scipy.stats.t.ppf для знаходження критичного значення

t
crit.

Зробіть висновок при рівні значущості α=0.05: чи є підстави стверджувати, що нова методика вплинула на результати?


Дані: x̄=76.8, s=14.2, n=35, μ0=72

1. Гіпотези

H0: μ=72  
H1: μ≠72  

2. Стандартна похибка

SE = s/√n = 14.2/√35 ≈ 2.40

3. t-статистика

t = (x̄ - μ0)/SE = (76.8 - 72)/2.40 = 2.00

4. p-value

Для df=34, t=2.00 → p ≈ 0.054 (двосторонній тест)

5. 95% довірчий інтервал

t_crit ≈ 2.03  
CI = 76.8 ± 2.03·2.40 = (71.9, 81.7)

6. Висновок.

Оскільки p ≈ 0.054 > 0.05, немає статистично значущої різниці.


```python
from scipy.stats import t
import numpy as np

xbar, s, n, mu0 = 76.8, 14.2, 35, 72
SE = s/np.sqrt(n)
t_stat = (xbar - mu0)/SE
df = n - 1

p_value = 2*(1 - t.cdf(abs(t_stat), df))
t_crit = t.ppf(1 - 0.025, df)
CI_low = xbar - t_crit*SE
CI_high = xbar + t_crit*SE

print("SE =", SE)
print("t =", t_stat)
print("p-value =", p_value)
print("95% CI =", (CI_low, CI_high))

if p_value < 0.05:
    print("Висновок: нова методика вплинула на результати.")
else:
    print("Висновок: статистично значущої різниці немає.")
```
