```python
import numpy as np

# Вектор користувача
u = np.array([8, 2, 5])

# Вектори фільмів
vA = np.array([9, 1, 2])
vB = np.array([1, 9, 8])
vC = np.array([7, 2, 6])

# Функція косинусної подібності
def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

# Обчислення схожості
sim_A = cosine_similarity(u, vA)
sim_B = cosine_similarity(u, vB)
sim_C = cosine_similarity(u, vC)

print("Фільм A:", sim_A)
print("Фільм B:", sim_B)
print("Фільм C:", sim_C)

# Найкращий фільм
similarities = {"A": sim_A, "B": sim_B, "C": sim_C}
best_movie = max(similarities, key=similarities.get)
print("Найбільш схожий фільм:", best_movie)
```

Задача 4. Прогноз тренду

При моніторингу IT-інфраструктури дані часто надходять із шумом.

Навантаження на CPU сервера може коливатися, але загальний тренд вказує на поступове зростання.

Щоб передбачити, коли сервер досягне критичного навантаження, потрібно побудувати лінію тренду

y=kx+b.



Оскільки точки не лежать ідеально на одній прямій, точного розв'язку системи не існує, тому використовується метод найменших квадратів (МНК).



Дано виміри завантаження CPU y протягом 5 годин t.

Час t=[1,2,3,4,5]

Навантаження y=[22,28,37,45,53]


Використовуючи бібліотеку numpy, напишіть програму та виконайте наступні кроки:

Сформуйте матрицю A. Пам'ятайте, що для знаходження вільного члена b матриця повинна містити стовпчик зі значень t та стовпчик з одиниць.

Знайдіть коефіцієнти прямої k та b за допомогою функції np.linalg.lstsq(A, y). Ця функція повертає найкращий наближений розв'язок.
Використовуючи знайдені коефіцієнти, обчисліть прогнозоване навантаження на 6-ту годину за формулою  
у =k⋅6+b.
Виведіть рівняння отриманого тренду та прогноз.

Опціональне завдання з зірочкою (+5 бонусних балів). Функція np.linalg.lstsq() використовує нормальне рівняння для знаходження оптимальних коефіцієнтів. Реалізуйте цей метод вручну.



Знайдіть коефіцієнти k - нахил, та b - зсув через нормальне рівняння  Ax=ATy. Де
x=[k,b]
T
  — вектор невідомих коефіцієнтів.



Кроки:

Обчислити ATA

Обчислити AT y
Розв'яжіть систему через np.linalg.solve()
Порівняйте результат з lstsq — вони мають збігатися

```python
import numpy as np

# Дані
t = np.array([1, 2, 3, 4, 5])
y = np.array([22, 28, 37, 45, 53])

# Матриця A (стовпчик t + стовпчик одиниць)
A = np.column_stack((t, np.ones(len(t))))

# Розв’язок через lstsq
coeffs, residuals, rank, s = np.linalg.lstsq(A, y, rcond=None)
k, b = coeffs
print("Рівняння тренду (через lstsq): y = {:.2f}x + {:.2f}".format(k, b))

# Прогноз на 6-ту годину
y_pred = k * 6 + b
print("Прогноз на 6-ту годину:", y_pred)

# Опціонально: нормальне рівняння
ATA = A.T @ A
ATy = A.T @ y
coeffs_normal = np.linalg.solve(ATA, ATy)
k_normal, b_normal = coeffs_normal
print("Рівняння тренду (нормальне рівняння): y = {:.2f}x + {:.2f}".format(k_normal, b_normal))
```
