Саме в цей час кількість активних користувацьких сесій найбільша.

```python
import numpy as np
from scipy.optimize import approx_fprime

# Функція
def f(t):
    return 1000 * t * np.exp(-0.2 * t)

# Аналітична похідна
def f_prime(t):
    return (1000 - 200 * t) * np.exp(-0.2 * t)

# Чисельна похідна
eps = np.sqrt(np.finfo(float).eps)
points = [2, 6, 10]

print("=== Задача 1 ===")
for t in points:
    num_deriv = approx_fprime(np.array([t]), f, eps)[0]
    print(f"t={t}, аналітична={f_prime(t):.3f}, чисельна={num_deriv:.3f}")

# Пікове навантаження
t_star = 1000 / 200
print(f"Максимум активності о {8+t_star:.2f} годині (≈13:00)")
```

Розв'язок.

# Дослідження впливу початкового рівня знань

Модель:
$$
\frac{dK}{dt} = r(M - K), \quad M=100, \quad r=0.15
$$

Загальний розв’язок:
$$
K(t) = M - (M - K(0))e^{-rt}
$$

---

## 1. Випадок K(0)=5
$$
K(t) = 100 - (100 - 5)e^{-0.15t} = 100 - 95e^{-0.15t}
$$

Знайдемо час досягнення 90%:
$$
100 - 95e^{-0.15t} = 90 \quad \Rightarrow \quad 10 = 95e^{-0.15t}
$$

$$
e^{-0.15t} = \frac{10}{95} \approx 0.1053
$$

$$
t = \frac{-\ln(0.1053)}{0.15} \approx 15.0 \text{ днів}
$$

---

## 2. Випадок K(0)=10
$$
K(t) = 100 - (100 - 10)e^{-0.15t} = 100 - 90e^{-0.15t}
$$

$$
100 - 90e^{-0.15t} = 90 \quad \Rightarrow \quad 10 = 90e^{-0.15t}
$$

$$
e^{-0.15t} = \frac{1}{9} \approx 0.1111
$$

$$
t = \frac{-\ln(0.1111)}{0.15} \approx 14.65 \text{ днів}
$$

---

## 3. Випадок K(0)=20
$$
K(t) = 100 - (100 - 20)e^{-0.15t} = 100 - 80e^{-0.15t}
$$

$$
100 - 80e^{-0.15t} = 90 \quad \Rightarrow \quad 10 = 80e^{-0.15t}
$$

$$
e^{-0.15t} = \frac{10}{80} = 0.125
$$

$$
t = \frac{-\ln(0.125)}{0.15} \approx 13.86 \text{ днів}
$$

---



Студенти з нижчим початковим рівнем (K(0)=5) досягає 90% знань приблизно за **15 днів**.  
Студент із **середнім рівнем (K(0)=10)** — за **14.65 днів**.  
Студент із **вищим рівнем (K(0)=20)** — за **13.86 днів**.  

Чим **вищий початковий рівень знань**, тим **швидше студент досягає високого рівня (90%)**. Початкова підготовка зменшує час навчання і робить процес ефективнішим.

```python
import numpy as np
from scipy.integrate import solve_ivp

def learning_rate(t, K):
    return 0.15 * (100 - K)

print("\n=== Задача 2 ===")
# Розв'язок для K0=10
sol = solve_ivp(learning_rate, [0, 30], [10], dense_output=True)
t_vals = np.linspace(0, 30, 100)
K_vals = sol.sol(t_vals)[0]

# Аналітична формула часу досягнення 90%
def time_to_90(K0):
    return -(1/0.15) * np.log(10 / (100 - K0))

for K0 in [5, 10, 20]:
    print(f"K0={K0}, час досягнення 90% ≈ {time_to_90(K0):.2f} днів")
```

$$
\frac{\partial f}{\partial x}(10,20) = 10 + 0.2\cdot20 + 10 = 24

$$
\frac{\partial f}{\partial y}(10,20) = 0.6\cdot20 + 0.2\cdot10 + 5 = 22
$$

Отже:
$$
\nabla f(10,20) = (24, 22)

\Delta x = 0.5, \quad \Delta y = -0.3

$$
\Delta f \approx \frac{\partial f}{\partial x}\cdot \Delta x + \frac{\partial f}{\partial y}\cdot \Delta y

Підставимо:
$$
\Delta f \approx 24\cdot0.5 + 22\cdot(-0.3) = 12 - 6.6 = 5.4

- Лінійна апроксимація добре наближає реальну зміну функції (похибка ≈1).  
Для бізнес‑аналітики це означає, що невеликі зміни параметрів $x$ і $y$ можна швидко оцінювати через градієнт, не обчислюючи всю функцію заново.




```python
import numpy as np
from scipy.optimize import approx_fprime

def f_xy(point):
    x, y = point
    return 0.5*x**2 + 0.3*y**2 + 0.2*x*y + 10*x + 5*y

print("\n=== Задача 4 ===")
# Аналітичні частинні похідні
def df_dx(x,y): return x + 0.2*y + 10
def df_dy(x,y): return 0.6*y + 0.2*x + 5

x0, y0 = 10, 20
print(f"Аналітичні похідні: df/dx={df_dx(x0,y0)}, df/dy={df_dy(x0,y0)}")

# Чисельний градієнт
eps = np.sqrt(np.finfo(float).eps)
grad = approx_fprime(np.array([x0,y0]), f_xy, eps)
print(f"Чисельний градієнт: {grad}")

# Лінійна апроксимація
dx, dy = 0.5, -0.3
delta_f_lin = df_dx(x0,y0)*dx + df_dy(x0,y0)*dy
delta_f_exact = f_xy([x0+dx, y0+dy]) - f_xy([x0,y0])
print(f"Δf (лінійна)={delta_f_lin:.3f}, Δf (точна)={delta_f_exact:.3f}")
```
